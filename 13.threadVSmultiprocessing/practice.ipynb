```python
import threading
import time

def print_message(name, count):
  for i in range(count):
      log.info(f"{name} says: {i}")
      time.sleep(1)


def main():
  thread = threading.Thread(target=print_message, args=("Threads", 6))
  thread.start()
  thread.join()
  log.info("Hello")

main()
```

```python
import threading
import time

def print_message(name, count):
  for i in range(count):
      log.info(f"{name} says: {i}")
      time.sleep(1)


def main():
  threads = []
  for i in range(3):
    thread = threading.Thread(target=print_message, args=(f"Thread {i+1}", 3))
    threads.append(thread)
    thread.start()

  for thread in threads:
    thread.join()
  log.info("Main thread already finished")

main()
```

```python
import threading
import time

def infinite_task():
  while True:
    log.info("Running ...")
    time.sleep(1)

def main():
  daemon_thread = threading.Thread(
    target=infinite_task
  )
  daemon_thread.daemon = True
  daemon_thread.start()
  time.sleep(1)
  log.info("Main thread finished")

main()
```
