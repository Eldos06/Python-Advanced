```python
import logging

logging.basicConfig(
  level = logging.INFO,
  datefmt="%H:%M:%S",
  format="[%(asctime)s.%(msecs)03d] %(lineno)3d %(funcName)15s %(levelname)-8s = %(message)s",
)

log = logging.getLogger(__name__)
```

```python
log.info("Hi")
```

```python
from time import sleep


def get_weather():
    log.info("fetching weather...")
    sleep(1)
    log.info("got weather")

def get_stocks():
    log.info("fetching stocks...")
    sleep(1)
    log.info("got stocks")
```

```python
def main():
    log.info("starting main...")
    get_weather()
    get_stocks()
    log.info("done main")

main()
```

```python
import threading

def main():
    log.info("starting main...")

    thread_weather = threading.Thread(target=get_weather)
    thread_stocks = threading.Thread(target=get_stocks)
    log.info("created thread")

    thread_weather.start()
    thread_stocks.start()
    log.info("created thread, will wait")

    thread_weather.join()
    thread_stocks.join()
    log.info("done thread")

    log.info("done main")

main()
```

```python
def get_stock_info(name):
    log.info("find stock %s", name)
    sleep(1)
    log.info("got stock %s", name)
    return {'name': name}
```

```python
from concurrent.futures import thread


def main():
  log.info("start main")

  thread_stock_a = threading.Thread(target=get_stock_info, args=("AAA", ))
  thread_stock_b =threading.Thread(target=get_stock_info, args=("BBB", ))
  log.info("created threads")

  thread_stock_a.start()
  thread_stock_b.start()
  log.info("started threads, will wait")

  thread_stock_a.join()
  thread_stock_b.join()
  log.info("finished thread")

  log.info("done main")


main()
```

```python

import concurrent.futures

def main():
  with concurrent.futures.ThreadPoolExecutor(max_workers=3) as executor:
    results = executor.map(get_stock_info, ["one", "two", "three", "four"])
  for res in results:
    log.info("result: %s", res)

main()
```

```python

import concurrent.futures

def main():
  with concurrent.futures.ThreadPoolExecutor(max_workers=3) as executor:
    future_stock = executor.submit(get_stock_info, "spam")

  for res in future_stock.result():
    log.info("result: %s", res)

main()
```

```python
timeout = 42_000_000

def countdown(n):
  log.info("countdown %s", n)
  while n:
    n -= 1
  log.info("done countdown")
```

```python
countdown(timeout)
```
